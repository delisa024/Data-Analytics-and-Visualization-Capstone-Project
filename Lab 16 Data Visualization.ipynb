**Read the CSV File into a Pandas DataFrame**

Load the Stack Overflow survey data into a Pandas DataFrame:


```python
# Read the CSV file
df = pd.read_csv('survey-data.csv')

# Display the first few rows of the data
pd.set_option("display.max_columns", None)
df.head()
```

**Create a SQLite Database and Insert the Data**

Now, let's create a new SQLite database (`survey-data.sqlite`) and insert the data from the DataFrame into a table using the sqlite3 library:


```python
import sqlite3

# Create a connection to the SQLite database
conn = sqlite3.connect('survey-data.sqlite')

# Write the dataframe to the SQLite database
df.to_sql('main', conn, if_exists='replace', index=False)


# Close the connection
conn.close()
```

**Verify the Data in the SQLite Database**
Verify that the data has been correctly inserted into the SQLite database by running a simple query:


```python
# Reconnect to the SQLite database
conn = sqlite3.connect('survey-data.sqlite')

# Run a simple query to check the data
QUERY = "SELECT * FROM main LIMIT 5"
df_check = pd.read_sql_query(QUERY, conn)

# Display the results
df_check
```

## Hands-on Lab


### Visualizing the Distribution of Data

**Histograms**

Plot a histogram of CompTotal (Total Compensation).


```python
df["CompTotal"]
```

```python
QUERY = """
SELECT Distinct CompTotal FROM main where CompTotal is not null
"""
df = pd.read_sql_query(QUERY, conn)
print(df.head())
```

```python
## Write your code here
df = pd.read_sql_query(
    "SELECT CompTotal FROM main WHERE CompTotal is not null",
    conn
)

# Convert to numeric
df["CompTotal"] = pd.to_numeric(df["CompTotal"], errors="coerce")

# Clean data and handle outliers
df = df.dropna()
df = df[df["CompTotal"] > 0]
df = df[df["CompTotal"] < df["CompTotal"].quantile(0.95)]

plt.hist(df, bins=50)
plt.xlabel("CompTotal")
plt.ylabel("Frequency")
plt.title("Histogram of Total Compensation")
plt.show()
```

**Box Plots**

Plot a box plot of Age.


```python
## Write your code here
QUERYAge = """
SELECT
    CASE
        WHEN Age = 'Under 18 years old' THEN 15
        WHEN Age = '18-24 years old' THEN 21
        WHEN Age = '25-34 years old' THEN 30
        WHEN Age = '35-44 years old' THEN 40
        WHEN Age = '45-54 years old' THEN 50
        WHEN Age = '55-64 years old' THEN 60
        WHEN Age = '65 years or older' THEN 70
        ELSE NULL
    END AS AgeNumeric
FROM main
WHERE Age IS NOT NULL
  AND Age != 'Prefer not to say'
"""
df = pd.read_sql_query(QUERYAge, conn)

plt.boxplot(df["AgeNumeric"])
plt.xlabel("Count")
plt.title("Boxplot of Age Counts")
plt.show()
```

### Visualizing Relationships in Data

**Scatter Plots**

Create a scatter plot of Age and WorkExp.


```python
## Write your code here
QUERYAge = """
SELECT
    CASE
        WHEN Age = 'Under 18 years old' THEN 15
        WHEN Age = '18-24 years old' THEN 21
        WHEN Age = '25-34 years old' THEN 30
        WHEN Age = '35-44 years old' THEN 40
        WHEN Age = '45-54 years old' THEN 50
        WHEN Age = '55-64 years old' THEN 60
        WHEN Age = '65 years or older' THEN 70
        ELSE NULL
    END AS Age,  WorkExp as WorkExp
FROM main
WHERE Age IS NOT NULL
  AND Age != 'Prefer not to say' AND WorkExp IS NOT NULL
"""
df = pd.read_sql_query(QUERYAge, conn)
plt.scatter(df["Age"], df["WorkExp"])
plt.xlabel("Age")
plt.ylabel("Work Experience")
plt.title("Scatter plot or Work Experience by Age")
plt.show()
```

**Bubble Plots**

Create a bubble plot of `TimeSearching` and `Frustration` using the Age column as the bubble size.


```python
QUERY = """
SELECT Frustration FROM main
WHERE Frustration IS NOT NULL
"""
df = pd.read_sql_query(QUERY, conn)

# Split + explode + clean
fr_split = df["Frustration"].str.split(";").explode().str.strip()

# Check values
print(fr_split.value_counts())
```

```python
## Write your code here
QUERY = """
SELECT
    CASE
        WHEN Age = 'Under 18 years old' THEN 15
        WHEN Age = '18-24 years old' THEN 21
        WHEN Age = '25-34 years old' THEN 30
        WHEN Age = '35-44 years old' THEN 40
        WHEN Age = '45-54 years old' THEN 50
        WHEN Age = '55-64 years old' THEN 60
        WHEN Age = '65 years or older' THEN 70
        ELSE NULL
    END AS Age,

    CASE
        WHEN TimeSearching = 'Less than 15 minutes a day' THEN 15
        WHEN TimeSearching = '15-30 minutes a day' THEN 30
        WHEN TimeSearching = '30-60 minutes a day' THEN 60
        WHEN TimeSearching = '60-120 minutes a day' THEN 120
        WHEN TimeSearching = 'Over 120 minutes a day' THEN 300
        ELSE NULL
    END AS TimeSearching,

    -- Count number of frustrations selected
    (LENGTH(Frustration) - LENGTH(REPLACE(Frustration, ';', '')) + 1) AS Frustration

FROM main
WHERE Age IS NOT NULL
  AND Age != 'Prefer not to say'
  AND TimeSearching IS NOT NULL
  AND Frustration != 'None of these'
"""
df = pd.read_sql_query(QUERY, conn)

df = df.dropna()

scatter = plt.scatter(
    df["TimeSearching"],
    df["Frustration"],   # now numeric (count of issues)
    s=df["Age"] * 5,
    c=df["Age"],
    cmap="viridis",
    alpha=0.5
)

plt.xlabel("Time Searching (minutes/day)")
plt.ylabel("Number of Frustrations")
plt.title("Time Searching vs Frustration Count (Bubble = Age)")

plt.colorbar(scatter, label="Age")

plt.show()
```

### Visualizing Composition of Data

**Pie Charts**

Create a pie chart of the top 5 databases(`DatabaseWantToWorkWith`) that respondents wish to learn next year.


```python
## Write your code here
df = pd.read_sql_query("""
SELECT DatabaseWantToWorkWith
FROM main
WHERE DatabaseWantToWorkWith IS NOT NULL
LIMIT 5
""", conn)

# Split ; separated values into individual rows
db_split = df["DatabaseWantToWorkWith"].str.split(";").explode()

# Remove extra spaces
db_split = db_split.str.strip()

# Count top 5 databases
top5 = db_split.value_counts().head(5)


plt.pie(
    top5.values,
    labels=top5.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Top 5 Databases People Want to Learn Next Year")

plt.show()
```

**Stacked Charts** 

Create a stacked bar chart of median `TimeSearching` and `TimeAnswering` for the age group 30 to 35.


```python
QUERY = """
SELECT Distinct TimeAnswering FROM main
WHERE TimeAnswering is not null
"""
pd.read_sql_query(QUERY, conn)
```

```python
QUERY = """
SELECT Distinct TimeSearching FROM main
WHERE TimeSearching is not null 
"""
pd.read_sql_query(QUERY, conn)
```

```python
## Write your code here
QUERY = """
SELECT
    CASE
        WHEN Age = '25-34 years old' THEN 30
        WHEN Age = '35-44 years old' THEN 40
        ELSE NULL
    END AS Age,

    CASE
        WHEN TimeSearching = '30-60 minutes a day' THEN 60
        WHEN TimeSearching = '15-30 minutes a day' THEN 30
        WHEN TimeSearching = 'Less than 15 minutes a day' THEN 15
        WHEN TimeSearching = '60-120 minutes a day' THEN 120
        WHEN TimeSearching = 'Over 120 minutes a day' THEN 300
        ELSE NULL
    END AS TimeSearching,

    CASE
        WHEN TimeAnswering = '30-60 minutes a day' THEN 60
        WHEN TimeAnswering = '15-30 minutes a day' THEN 30
        WHEN TimeAnswering = 'Less than 15 minutes a day' THEN 15
        WHEN TimeAnswering = '60-120 minutes a day' THEN 120
        WHEN TimeAnswering = 'Over 120 minutes a day' THEN 300
        ELSE NULL
    END AS TimeAnswering

FROM main
WHERE Age IN ('25-34 years old', '35-44 years old')
"""
df = pd.read_sql_query(QUERY, conn)
df = df.dropna()
df = df[df["Age"] == 30]
median_search = df["TimeSearching"].median()
median_answer = df["TimeAnswering"].median()

# Data
labels = ["Age 30–35"]
search_vals = [median_search]
answer_vals = [median_answer]

# Plot
plt.bar(labels, search_vals, label="Time Searching")
plt.bar(labels, answer_vals, bottom=search_vals, label="Time Answering")

plt.ylabel("Median Time (encoded scale)")
plt.title("Median Time Searching vs Answering (Age 30–35)")
plt.legend()

plt.show()
```

### Visualizing Comparison of Data

**Line Chart**

Plot the median `CompTotal` for all ages from 45 to 60.


```python
## Write your code here
QUERY = """
SELECT
    CASE
        WHEN Age = '45-54 years old' THEN 50
        WHEN Age = '55-64 years old' THEN 60
        ELSE NULL
    END AS Age,

    CompTotal

FROM main
WHERE Age IN ('45-54 years old', '55-64 years old')
  AND CompTotal IS NOT NULL
"""
df = pd.read_sql_query(QUERY, conn)

# Convert to numeric
df["CompTotal"] = pd.to_numeric(df["CompTotal"], errors="coerce")

# Clean data, handle outliers and calculate median by age
df = df.dropna()
df = df[df["CompTotal"] > 0]
df = df[df["CompTotal"] < df["CompTotal"].quantile(0.95)]
median_df = df.groupby("Age")["CompTotal"].median().reset_index()

plt.plot(median_df["Age"], median_df["CompTotal"], marker='o')

plt.xlabel("Age")
plt.ylabel("Median Compensation")
plt.title("Median Compensation for Ages 45–60")

plt.grid(True)

plt.show()
```

**Bar Chart**

Create a horizontal bar chart using the `MainBranch` column.


```python
## Write your code here
df = pd.read_sql_query("""
SELECT MainBranch
FROM main
WHERE MainBranch IS NOT NULL
""", conn)
## count values
counts = df["MainBranch"].value_counts()

counts.plot(kind="barh")

plt.xlabel("Count")
plt.ylabel("Main Branch")
plt.title("Distribution of Main Branch")

plt.show()
```
